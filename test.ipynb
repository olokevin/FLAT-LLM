```python
from graphviz import Digraph
import os

# Ensure output directory exists
os.makedirs('./figs', exist_ok=True)

dot = Digraph('CompressThenTrain', format='png')
dot.attr(rankdir='TB', bgcolor='white')  # Top to Bottom layout

# Define common node styles
node_style = {
    'style': 'filled',
    'fontname': 'Helvetica',
    'fontsize': '11',
    'fillcolor': '#e0f7fa',
    'shape': 'box',
    'color': '#26a69a'
}

# Nodes
dot.node('A', 'Compress-then-train', shape='oval', style='filled', fillcolor='#b2dfdb', fontname='Helvetica')

dot.node('B', '1. Post-training compression:\nShortlist candidates\nmeeting target compression ratio', **node_style)
dot.node('C', '2. Lightweight training:\nEvaluate candidates on\nperplexity / downstream task', **node_style)
dot.node('D', '3. Re-training with\nknowledge distillation', **node_style)

# Edges
dot.edge('A', 'B')
dot.edge('B', 'C')
dot.edge('C', 'D')

# Render without viewing
dot.render('./figs/compress_then_train', view=False)
```
